```python
team_runs = df.groupby("batting_team")["runs_total"].sum().sort_values(ascending=False)

team_runs.head(10)
```

```python
top_batters = df.groupby("batter")["batter_runs"].sum().sort_values(ascending=False).head(10)

top_batters
```

```python
top_bowlers = df.groupby("bowler")["bowler_wicket"].sum().sort_values(ascending=False).head(10)

top_bowlers
```

```python
runs_by_over = df.groupby("over")["runs_total"].sum()

runs_by_over
```

```python
import matplotlib.pyplot as plt

top_batters.plot(kind="bar")

plt.title("Top IPL Run Scorers")
plt.xlabel("Player")
plt.ylabel("Runs")

plt.show()
```

```python
from sqlalchemy import create_engine

engine = create_engine("sqlite:///ipl.db")

df.to_sql("ipl_data", engine, if_exists="replace", index=False)
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("ipl.db")

query = "SELECT * FROM ipl_data LIMIT 5"

pd.read_sql(query, conn)
```

```python
query = """
SELECT batter, SUM(batter_runs) as total_runs
FROM ipl_data
GROUP BY batter
ORDER BY total_runs DESC
LIMIT 10
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT bowler, SUM(bowler_wicket) as wickets
FROM ipl_data
GROUP BY bowler
ORDER BY wickets DESC
LIMIT 10
"""

pd.read_sql(query, conn)
```
